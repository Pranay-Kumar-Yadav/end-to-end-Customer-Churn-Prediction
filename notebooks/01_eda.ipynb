# Customer Churn: Exploratory Analysis
Load the source CSV and inspect its schema, quality, and churn balance.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

DATA_PATH = Path('..') / 'data' / 'raw' / 'customer_churn.csv'
data = pd.read_csv(DATA_PATH)
data.head()
```

```python
data.shape, data.dtypes, data.isna().sum().sort_values(ascending=False).head(10)
```

```python
if 'Churn' not in data or data.empty:
    raise ValueError('Replace the starter header-only CSV with labeled observations first.')
ax = sns.countplot(data=data, x='Churn')
ax.set_title('Customer churn balance')
plt.tight_layout()
```
